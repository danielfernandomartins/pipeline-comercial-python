```python
oportunidades = []


def cadastrar_oportunidade():
    cliente = input("Nome do cliente: ")
    produto = input("Produto ou serviço: ")
    valor = float(input("Valor da oportunidade: R$ "))
    status = input(
        "Status (prospeccao / negociacao / ganha / perdida): "
    ).lower()

    oportunidade = {
        "cliente": cliente,
        "produto": produto,
        "valor": valor,
        "status": status
    }

    oportunidades.append(oportunidade)

    print("\nOportunidade cadastrada com sucesso!")


def listar_oportunidades():

    if len(oportunidades) == 0:
        print("\nNenhuma oportunidade cadastrada.")
        return

    print("\n=== PIPELINE COMERCIAL ===")

    for indice, oportunidade in enumerate(oportunidades, start=1):

        print(
            f"{indice} - "
            f"Cliente: {oportunidade['cliente']} | "
            f"Produto: {oportunidade['produto']} | "
            f"Valor: R$ {oportunidade['valor']:.2f} | "
            f"Status: {oportunidade['status']}"
        )


def mostrar_pipeline():

    total_pipeline = 0

    for oportunidade in oportunidades:

        if oportunidade["status"] in ["prospeccao", "negociacao"]:
            total_pipeline += oportunidade["valor"]

    print(
        f"\nValor total das oportunidades em aberto: "
        f"R$ {total_pipeline:.2f}"
    )


def calcular_resultados():

    total = len(oportunidades)

    if total == 0:
        print("\nNenhuma oportunidade cadastrada.")
        return

    ganhas = 0
    perdidas = 0
    valor_ganho = 0

    for oportunidade in oportunidades:

        if oportunidade["status"] == "ganha":
            ganhas += 1
            valor_ganho += oportunidade["valor"]

        elif oportunidade["status"] == "perdida":
            perdidas += 1

    taxa_conversao = (ganhas / total) * 100

    print("\n=== RESULTADOS COMERCIAIS ===")

    print(f"Total de oportunidades: {total}")
    print(f"Oportunidades ganhas: {ganhas}")
    print(f"Oportunidades perdidas: {perdidas}")
    print(f"Valor das vendas ganhas: R$ {valor_ganho:.2f}")
    print(f"Taxa de conversão: {taxa_conversao:.2f}%")


while True:

    print("\n================================")
    print("       PIPELINE COMERCIAL")
    print("================================")

    print("1 - Cadastrar oportunidade")
    print("2 - Listar oportunidades")
    print("3 - Mostrar valor do pipeline")
    print("4 - Mostrar resultados")
    print("5 - Sair")

    opcao = input("\nEscolha uma opção: ")

    if opcao == "1":
        cadastrar_oportunidade()

    elif opcao == "2":
        listar_oportunidades()

    elif opcao == "3":
        mostrar_pipeline()

    elif opcao == "4":
        calcular_resultados()

    elif opcao == "5":
        print("\nPrograma encerrado.")
        break

    else:
        print("\nOpção inválida.")
```
